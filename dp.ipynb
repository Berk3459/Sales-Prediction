```python
from sklearn.preprocessing import MinMaxScaler
from keras.models import Sequential
from keras.layers import LSTM,Dense
```

```python
#zaman serisi modeli için gecikmeli değerleri oluşturalım
def ts(veri, timestep):
    x, y = [], []
    for i in range(timestep, len(veri)):
        x.append(veri.iloc[i-timestep:i, 0].values)
        y.append(veri.iloc[i, 0])
    return np.array(x), np.array(y)
```

```python
x_train,y_train = ts(train,2)
x_test,y_test = ts(test,2)
```

```python
x_train = np.reshape(x_train,(x_train.shape[0],1,x_train.shape[1]))
x_test = np.reshape(x_test,(x_test.shape[0],1,x_test.shape[1]))
y_train = y_train.reshape(-1,1)
y_test = y_test.reshape(-1,1)
```

```python
#modelimizi kuralım
model = Sequential()
model.add(LSTM(16,input_shape = (x_train.shape[1],2)))
model.add(Dense(1))
model.compile(optimizer="adam",loss="mse")
model.fit(x_train,y_train,epochs=50)
model.summary()
```

```python
#tahmin değerlerini oluşturalım
train_tahmin = model.predict(x_train)
test_tahmin = model.predict(x_test)
```
